```python
# ============================================================
# V9 — DOWNLOAD ESA WORLDCOVER 2021 V200
# INDIA TILES ONLY
# ============================================================

import os
import requests
from concurrent.futures import ThreadPoolExecutor, as_completed

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

OUTPUT_DIR = "worldcover_india"

os.makedirs(OUTPUT_DIR, exist_ok=True)

BASE_URL = (
    "https://esa-worldcover.s3.eu-central-1.amazonaws.com/"
    "v200/2021/map/"
)

# India bounding region
# Slightly expanded so border areas are included
MIN_LAT = 6
MAX_LAT = 39
MIN_LON = 66
MAX_LON = 99

TILE_SIZE = 3

# ------------------------------------------------------------
# GENERATE TILES
# ------------------------------------------------------------

tiles = []

for lat in range(MIN_LAT, MAX_LAT, TILE_SIZE):
    for lon in range(MIN_LON, MAX_LON, TILE_SIZE):

        # WorldCover naming convention
        lat_prefix = "N" if lat >= 0 else "S"
        lon_prefix = "E" if lon >= 0 else "W"

        tile = (
            f"{lat_prefix}{abs(lat):02d}"
            f"{lon_prefix}{abs(lon):03d}"
        )

        filename = (
            f"ESA_WorldCover_10m_2021_v200_"
            f"{tile}_Map.tif"
        )

        url = BASE_URL + filename

        tiles.append({
            "tile": tile,
            "filename": filename,
            "url": url
        })

print("=" * 70)
print("ESA WORLDCOVER INDIA DOWNLOAD")
print("=" * 70)

print(f"Candidate tiles : {len(tiles)}")
print(f"Output folder   : {os.path.abspath(OUTPUT_DIR)}")

print("\nFirst few tiles:")
for x in tiles[:10]:
    print(" ", x["tile"])

# ------------------------------------------------------------
# DOWNLOAD FUNCTION
# ------------------------------------------------------------

def download_tile(tile_info):

    tile = tile_info["tile"]
    filename = tile_info["filename"]
    url = tile_info["url"]

    output_path = os.path.join(
        OUTPUT_DIR,
        filename
    )

    # Skip if already downloaded
    if os.path.exists(output_path):

        size_mb = os.path.getsize(output_path) / (1024 ** 2)

        return (
            tile,
            "EXISTS",
            size_mb
        )

    try:

        response = requests.get(
            url,
            stream=True,
            timeout=120
        )

        if response.status_code != 200:

            return (
                tile,
                f"HTTP {response.status_code}",
                0
            )

        total_bytes = 0

        with open(output_path, "wb") as f:

            for chunk in response.iter_content(
                chunk_size=1024 * 1024
            ):

                if chunk:

                    f.write(chunk)
                    total_bytes += len(chunk)

        size_mb = total_bytes / (1024 ** 2)

        return (
            tile,
            "DOWNLOADED",
            size_mb
        )

    except Exception as e:

        return (
            tile,
            f"ERROR: {str(e)[:100]}",
            0
        )

# ------------------------------------------------------------
# DOWNLOAD
# ------------------------------------------------------------

print("\nStarting downloads...\n")

results = []

with ThreadPoolExecutor(max_workers=4) as executor:

    futures = [
        executor.submit(download_tile, tile)
        for tile in tiles
    ]

    for future in as_completed(futures):

        result = future.result()

        results.append(result)

        tile, status, size = result

        print(
            f"{tile:<10} "
            f"{status:<20} "
            f"{size:>8.2f} MB"
        )

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DOWNLOAD SUMMARY")
print("=" * 70)

downloaded = [
    r for r in results
    if r[1] in ["DOWNLOADED", "EXISTS"]
]

failed = [
    r for r in results
    if r[1] not in ["DOWNLOADED", "EXISTS"]
]

print(f"Successful : {len(downloaded)}")
print(f"Failed     : {len(failed)}")

if failed:

    print("\nFailed tiles:")

    for r in failed:
        print(
            f"  {r[0]} -> {r[1]}"
        )

print(
    f"\nTotal downloaded size: "
    f"{sum(r[2] for r in downloaded):.2f} MB"
)

print(
    f"\nFiles saved to:\n"
    f"{os.path.abspath(OUTPUT_DIR)}"
)
```

```python
# ============================================================
# V9 — CELL 4
# EXTRACT WORLDCOVER CLASS AT EVENT CENTROIDS
# ============================================================

import os
import glob
import rasterio
import numpy as np
import pandas as pd

WORLD_COVER_DIR = "worldcover_india"

# ------------------------------------------------------------
# Load downloaded WorldCover tiles
# ------------------------------------------------------------

raster_files = sorted(
    glob.glob(
        os.path.join(
            WORLD_COVER_DIR,
            "ESA_WorldCover_10m_2021_v200_*_Map.tif"
        )
    )
)

print("=" * 70)
print("WORLDCOVER EVENT EXTRACTION")
print("=" * 70)

print(f"WorldCover tiles : {len(raster_files)}")
print(f"VIIRS events     : {len(events):,}")

# ------------------------------------------------------------
# Build spatial index from raster bounds
# ------------------------------------------------------------

tile_info = []

for path in raster_files:

    with rasterio.open(path) as src:

        tile_info.append({
            "path": path,
            "left": src.bounds.left,
            "right": src.bounds.right,
            "bottom": src.bounds.bottom,
            "top": src.bounds.top
        })

print("\nTile index created.")

# ------------------------------------------------------------
# WorldCover class extraction
# ------------------------------------------------------------

# ESA WorldCover classes:
#
# 10  Tree cover
# 20  Shrubland
# 30  Grassland
# 40  Cropland
# 50  Built-up
# 60  Bare / sparse vegetation
# 70  Snow and ice
# 80  Permanent water bodies
# 90  Herbaceous wetland
# 95  Mangroves
# 100 Moss and lichen

def find_tile(lat, lon):

    for tile in tile_info:

        if (
            tile["left"] <= lon < tile["right"]
            and
            tile["bottom"] <= lat < tile["top"]
        ):
            return tile["path"]

    return None


landcover_values = []
landcover_tile = []

for idx, row in events.iterrows():

    lat = row["centroid_lat"]
    lon = row["centroid_lon"]

    path = find_tile(lat, lon)

    if path is None:

        landcover_values.append(np.nan)
        landcover_tile.append(None)
        continue

    try:

        with rasterio.open(path) as src:

            value = next(
                src.sample([(lon, lat)])
            )[0]

        landcover_values.append(int(value))
        landcover_tile.append(os.path.basename(path))

    except Exception:

        landcover_values.append(np.nan)
        landcover_tile.append(None)

# ------------------------------------------------------------
# Add results
# ------------------------------------------------------------

events["landcover_class"] = landcover_values
events["landcover_tile"] = landcover_tile

# ------------------------------------------------------------
# Class names
# ------------------------------------------------------------

class_names = {
    10: "Tree cover",
    20: "Shrubland",
    30: "Grassland",
    40: "Cropland",
    50: "Built-up",
    60: "Bare/sparse vegetation",
    70: "Snow/ice",
    80: "Permanent water",
    90: "Herbaceous wetland",
    95: "Mangroves",
    100: "Moss/lichen"
}

events["landcover_name"] = (
    events["landcover_class"]
    .map(class_names)
)

# ------------------------------------------------------------
# Diagnostics
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("EXTRACTION RESULTS")
print("=" * 70)

valid = events["landcover_class"].notna().sum()
missing = events["landcover_class"].isna().sum()

print(f"Events with land cover : {valid:,} ({valid/len(events)*100:.2f}%)")
print(f"Missing                 : {missing:,} ({missing/len(events)*100:.2f}%)")

print("\nLand-cover classes:")
print(
    events["landcover_name"]
    .value_counts(dropna=False)
    .to_string()
)

print("\nClass codes:")
print(
    events["landcover_class"]
    .value_counts(dropna=False)
    .sort_index()
    .to_string()
)

print("\nSample:")
print(
    events[
        [
            "event_id",
            "centroid_lat",
            "centroid_lon",
            "landcover_class",
            "landcover_name"
        ]
    ]
    .head(20)
    .to_string(index=False)
)
```

```python
# ============================================================
# V9 — CELL 5
# ROBUST WORLDCOVER COMPOSITION EXTRACTION
# ============================================================

import os
import glob
import rasterio
import numpy as np
import pandas as pd
from collections import defaultdict

WORLD_COVER_DIR = "worldcover_india"

RADIUS_300M = 300
RADIUS_1KM = 1000
PIXEL_SIZE_M = 10

WINDOW_300M = RADIUS_300M // PIXEL_SIZE_M
WINDOW_1KM = RADIUS_1KM // PIXEL_SIZE_M

print("=" * 70)
print("ROBUST WORLDCOVER COMPOSITION EXTRACTION")
print("=" * 70)

print(f"Events       : {len(events):,}")
print(f"300 m radius : {RADIUS_300M} m")
print(f"1 km radius  : {RADIUS_1KM} m")


# ------------------------------------------------------------
# 1. Find WorldCover tiles
# ------------------------------------------------------------

raster_files = sorted(
    glob.glob(
        os.path.join(
            WORLD_COVER_DIR,
            "ESA_WorldCover_10m_2021_v200_*_Map.tif"
        )
    )
)

print(f"\nWorldCover files found : {len(raster_files)}")


# ------------------------------------------------------------
# 2. Validate tiles
# ------------------------------------------------------------

valid_tiles = []
bad_tiles = []

print("\nValidating tiles...")

for i, path in enumerate(raster_files, 1):

    try:
        with rasterio.open(path) as src:

            # Read a small block from the beginning.
            # This catches many corrupt TIFFs without
            # loading the entire 10 m raster.
            src.read(
                1,
                window=rasterio.windows.Window(
                    0, 0,
                    min(100, src.width),
                    min(100, src.height)
                )
            )

            valid_tiles.append({
                "path": path,
                "left": src.bounds.left,
                "right": src.bounds.right,
                "bottom": src.bounds.bottom,
                "top": src.bounds.top,
                "width": src.width,
                "height": src.height
            })

    except Exception as e:

        bad_tiles.append({
            "path": path,
            "error": str(e)
        })

    print(
        f"\rChecked {i}/{len(raster_files)}",
        end=""
    )

print("\n")

print(f"Valid tiles : {len(valid_tiles)}")
print(f"Bad tiles   : {len(bad_tiles)}")

if bad_tiles:

    print("\nBad/corrupt tiles:")

    for tile in bad_tiles:
        print(
            os.path.basename(tile["path"])
        )


# ------------------------------------------------------------
# 3. Build tile lookup
# ------------------------------------------------------------

def find_tile(lat, lon):

    for tile in valid_tiles:

        if (
            tile["left"] <= lon < tile["right"]
            and
            tile["bottom"] <= lat < tile["top"]
        ):
            return tile["path"]

    return None


events["wc_tile_path"] = [
    find_tile(
        row["centroid_lat"],
        row["centroid_lon"]
    )
    for _, row in events.iterrows()
]

valid_events = events[
    events["wc_tile_path"].notna()
].copy()

print("\nEvents assigned to valid tiles:")
print(
    f"{len(valid_events):,} / {len(events):,}"
)


# ------------------------------------------------------------
# 4. Group events by tile
# ------------------------------------------------------------

events_by_tile = defaultdict(list)

for idx, row in valid_events.iterrows():

    events_by_tile[
        row["wc_tile_path"]
    ].append(idx)

print(
    f"Tiles containing events : "
    f"{len(events_by_tile)}"
)


# ------------------------------------------------------------
# 5. WorldCover classes
# ------------------------------------------------------------

class_codes = [
    10,
    20,
    30,
    40,
    50,
    60,
    70,
    80,
    90,
    95,
    100
]

class_names = {
    10: "tree",
    20: "shrub",
    30: "grass",
    40: "crop",
    50: "builtup",
    60: "bare",
    70: "snow",
    80: "water",
    90: "wetland",
    95: "mangrove",
    100: "moss"
}


# ------------------------------------------------------------
# 6. Output storage
# ------------------------------------------------------------

composition_features = {}

for idx in events.index:

    composition_features[idx] = {}


# ------------------------------------------------------------
# 7. Process tiles
# ------------------------------------------------------------

failed_reads = []

tile_count = len(events_by_tile)

for tile_num, (tile_path, event_indices) in enumerate(
    events_by_tile.items(),
    1
):

    print(
        f"\rProcessing tile "
        f"{tile_num}/{tile_count} "
        f"| Events: {len(event_indices):4d}",
        end=""
    )

    try:

        with rasterio.open(tile_path) as src:

            for idx in event_indices:

                row = events.loc[idx]

                lat = row["centroid_lat"]
                lon = row["centroid_lon"]

                try:

                    col, raster_row = src.index(
                        lon,
                        lat
                    )

                    # ------------------------------------------------
                    # 300 m window
                    # ------------------------------------------------

                    r0 = max(
                        0,
                        raster_row - WINDOW_300M
                    )

                    r1 = min(
                        src.height,
                        raster_row + WINDOW_300M + 1
                    )

                    c0 = max(
                        0,
                        col - WINDOW_300M
                    )

                    c1 = min(
                        src.width,
                        col + WINDOW_300M + 1
                    )

                    window_300 = src.read(
                        1,
                        window=rasterio.windows.Window(
                            c0,
                            r0,
                            c1 - c0,
                            r1 - r0
                        )
                    )


                    # ------------------------------------------------
                    # 1 km window
                    # ------------------------------------------------

                    r0_1 = max(
                        0,
                        raster_row - WINDOW_1KM
                    )

                    r1_1 = min(
                        src.height,
                        raster_row + WINDOW_1KM + 1
                    )

                    c0_1 = max(
                        0,
                        col - WINDOW_1KM
                    )

                    c1_1 = min(
                        src.width,
                        col + WINDOW_1KM + 1
                    )

                    window_1km = src.read(
                        1,
                        window=rasterio.windows.Window(
                            c0_1,
                            r0_1,
                            c1_1 - c0_1,
                            r1_1 - r0_1
                        )
                    )


                    # ------------------------------------------------
                    # Calculate percentages
                    # ------------------------------------------------

                    feature_dict = {}

                    for radius_name, data in [
                        ("300m", window_300),
                        ("1km", window_1km)
                    ]:

                        total_pixels = data.size

                        for code in class_codes:

                            percentage = (
                                np.sum(data == code)
                                / total_pixels
                                * 100
                            )

                            feature_name = (
                                f"{class_names[code]}"
                                f"_pct_{radius_name}"
                            )

                            feature_dict[
                                feature_name
                            ] = percentage

                    composition_features[idx] = feature_dict

                except Exception as e:

                    failed_reads.append({
                        "event_index": idx,
                        "tile": os.path.basename(tile_path),
                        "error": str(e)
                    })

    except Exception as e:

        print(
            f"\nFAILED TILE: "
            f"{os.path.basename(tile_path)}"
        )

        failed_reads.append({
            "event_index": None,
            "tile": os.path.basename(tile_path),
            "error": str(e)
        })


print("\n\nComposition extraction complete.")


# ------------------------------------------------------------
# 8. Convert to DataFrame
# ------------------------------------------------------------

composition_df = pd.DataFrame.from_dict(
    composition_features,
    orient="index"
)

composition_df.index.name = "event_index"


# ------------------------------------------------------------
# 9. Merge into events
# ------------------------------------------------------------

events = events.join(
    composition_df,
    how="left"
)


# ------------------------------------------------------------
# 10. Remove temporary tile column
# ------------------------------------------------------------

events.drop(
    columns=["wc_tile_path"],
    inplace=True
)


# ------------------------------------------------------------
# 11. Diagnostics
# ------------------------------------------------------------

landcover_feature_columns = [
    c for c in events.columns
    if "_pct_300m" in c
    or "_pct_1km" in c
]

complete_landcover = events[
    landcover_feature_columns
].notna().all(axis=1).sum()

print("\n" + "=" * 70)
print("LAND-COVER COMPOSITION RESULTS")
print("=" * 70)

print(
    f"Events with composition : "
    f"{complete_landcover:,}"
)

print(
    f"Events without composition : "
    f"{len(events) - complete_landcover:,}"
)

print(
    f"Land-cover features : "
    f"{len(landcover_feature_columns)}"
)

print(
    f"Failed reads : "
    f"{len(failed_reads)}"
)

print("\nSample features:")

display(
    events[
        [
            "event_id",
            "landcover_class",
            "landcover_name"
        ] +
        landcover_feature_columns[:10]
    ].head(10)
)
```

```python
# ============================================================
# V9 — CELL 6
# FINAL EVENT FEATURE TABLE
# ============================================================

print("=" * 70)
print("FINAL EVENT FEATURE TABLE")
print("=" * 70)

# ------------------------------------------------------------
# 1. Check current dataset
# ------------------------------------------------------------

print(f"Events : {len(events):,}")
print(f"Columns before cleanup : {len(events.columns)}")


# ------------------------------------------------------------
# 2. Remove non-ML metadata columns
# ------------------------------------------------------------

metadata_columns = [
    "start_date",
    "end_date",
    "landcover_name",
    "landcover_tile"
]

# Only remove columns that actually exist
metadata_columns = [
    c for c in metadata_columns
    if c in events.columns
]

events_ml = events.drop(
    columns=metadata_columns,
    errors="ignore"
).copy()


# ------------------------------------------------------------
# 3. Identify feature groups
# ------------------------------------------------------------

thermal_columns = [
    "mean_frp",
    "max_frp",
    "std_frp",
    "mean_bright_ti4",
    "max_bright_ti4",
    "std_bright_ti4",
    "mean_bright_ti5",
    "max_bright_ti5",
    "std_bright_ti5",
    "frp_range",
    "ti4_range",
    "ti5_range"
]

temporal_columns = [
    "active_days",
    "detection_count",
    "duration_days",
    "activity_frequency",
    "detections_per_active_day"
]

spatial_columns = [
    "spatial_extent_km"
]

osm_columns = [
    c for c in events_ml.columns
    if (
        "osm" in c
        or "industrial_" in c
        or "_count_1km" in c
        or "_count_3km" in c
    )
]

landcover_columns = [
    c for c in events_ml.columns
    if "_pct_300m" in c
    or "_pct_1km" in c
]


# ------------------------------------------------------------
# 4. Verify columns
# ------------------------------------------------------------

print("\nFeature groups:")

print(f"Thermal    : {len(thermal_columns)}")
print(f"Temporal   : {len(temporal_columns)}")
print(f"Spatial    : {len(spatial_columns)}")
print(f"OSM        : {len(osm_columns)}")
print(f"Land cover : {len(landcover_columns)}")


# ------------------------------------------------------------
# 5. ML feature list
# ------------------------------------------------------------

feature_columns = (
    thermal_columns
    + temporal_columns
    + spatial_columns
    + osm_columns
    + landcover_columns
)

# Keep only columns actually present
feature_columns = [
    c for c in feature_columns
    if c in events_ml.columns
]

print(
    f"\nTotal candidate ML features : "
    f"{len(feature_columns)}"
)


# ------------------------------------------------------------
# 6. Missing-value analysis
# ------------------------------------------------------------

feature_missing = (
    events_ml[feature_columns]
    .isna()
    .sum()
    .sort_values(ascending=False)
)

print("\nFeatures with missing values:")

print(
    feature_missing[
        feature_missing > 0
    ].to_string()
)


# ------------------------------------------------------------
# 7. Create final ML matrix
# ------------------------------------------------------------

X_raw = events_ml[
    feature_columns
].copy()

# Numeric conversion
X_raw = X_raw.apply(
    pd.to_numeric,
    errors="coerce"
)

print("\nRaw feature matrix:")
print(f"Rows    : {X_raw.shape[0]:,}")
print(f"Features: {X_raw.shape[1]}")


# ------------------------------------------------------------
# 8. Handle missing values
# ------------------------------------------------------------

# Land-cover missingness is caused by the two corrupt/
# unavailable WorldCover tiles.
#
# For clustering, missing contextual percentages are
# represented as zero because those events have no
# available land-cover context.

X_raw = X_raw.fillna(0)


# ------------------------------------------------------------
# 9. Sanity check
# ------------------------------------------------------------

print("\nMissing values after imputation:")
print(
    X_raw.isna().sum().sum()
)

print("\nInfinite values:")
print(
    np.isinf(X_raw.to_numpy()).sum()
)


# ------------------------------------------------------------
# 10. Final feature summary
# ------------------------------------------------------------

feature_summary = pd.DataFrame({
    "feature": X_raw.columns,
    "min": X_raw.min().values,
    "max": X_raw.max().values,
    "mean": X_raw.mean().values,
    "std": X_raw.std().values
})

print("\nFeature summary:")
display(feature_summary)


# ------------------------------------------------------------
# 11. Save intermediate dataset
# ------------------------------------------------------------

events_ml[
    ["event_id"] + feature_columns
].to_csv(
    "viirs_event_feature_table_v9.csv",
    index=False
)

print(
    "\nSaved:"
    "\nviirs_event_feature_table_v9.csv"
)
```

```python
# ============================================================
# V10 — CELL 8
# ROBUST SCALING + K-MEANS MODEL COMPARISON
# ============================================================

from sklearn.preprocessing import RobustScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import pandas as pd
import numpy as np
import time

print("=" * 70)
print("K-MEANS CLUSTERING")
print("=" * 70)

print(f"Events   : {X_cluster.shape[0]:,}")
print(f"Features : {X_cluster.shape[1]}")


# ------------------------------------------------------------
# 1. Robust scaling
# ------------------------------------------------------------

scaler = RobustScaler()

X_scaled = scaler.fit_transform(
    X_cluster
)

print("\nScaling complete.")

print(
    f"Scaled matrix shape : "
    f"{X_scaled.shape}"
)


# ------------------------------------------------------------
# 2. Test multiple K values
# ------------------------------------------------------------

k_values = range(2, 9)

results = []

print("\nTesting K = 2 to 8...\n")

for k in k_values:

    start_time = time.time()

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = model.fit_predict(
        X_scaled
    )

    silhouette = silhouette_score(
        X_scaled,
        labels
    )

    elapsed = time.time() - start_time

    results.append({
        "k": k,
        "inertia": model.inertia_,
        "silhouette_score": silhouette,
        "time_seconds": elapsed
    })

    print(
        f"K={k} | "
        f"Inertia={model.inertia_:,.2f} | "
        f"Silhouette={silhouette:.4f} | "
        f"Time={elapsed:.2f}s"
    )


# ------------------------------------------------------------
# 3. Results table
# ------------------------------------------------------------

k_results = pd.DataFrame(results)

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

display(
    k_results
)


# ------------------------------------------------------------
# 4. Best silhouette
# ------------------------------------------------------------

best_silhouette_row = k_results.loc[
    k_results["silhouette_score"].idxmax()
]

print(
    "\nBest silhouette K : "
    f"{int(best_silhouette_row['k'])}"
)

print(
    "Best silhouette   : "
    f"{best_silhouette_row['silhouette_score']:.4f}"
)


# ------------------------------------------------------------
# 5. Save scaling/model-selection information
# ------------------------------------------------------------

k_results.to_csv(
    "kmeans_model_comparison_v10.csv",
    index=False
)

print(
    "\nSaved:"
    "\nkmeans_model_comparison_v10.csv"
)
```

```python
# ============================================================
# V10 — CELL 10
# BEHAVIOR-ONLY K-MEANS
# ============================================================

from sklearn.preprocessing import RobustScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import pandas as pd
import numpy as np

print("=" * 70)
print("BEHAVIOR-ONLY CLUSTERING")
print("=" * 70)


# ------------------------------------------------------------
# 1. Behavior features
# ------------------------------------------------------------

behavior_features = [
    # Thermal intensity
    "mean_frp",
    "max_frp",
    "mean_bright_ti4",
    "max_bright_ti4",
    "mean_bright_ti5",
    "max_bright_ti5",

    # Temporal persistence
    "active_days",
    "duration_days",
    "activity_frequency",
    "detections_per_active_day",

    # Spatial behavior
    "spatial_extent_km"
]

behavior_features = [
    c for c in behavior_features
    if c in X_raw.columns
]

X_behavior = X_raw[
    behavior_features
].copy()


# ------------------------------------------------------------
# 2. Scale
# ------------------------------------------------------------

behavior_scaler = RobustScaler()

X_behavior_scaled = behavior_scaler.fit_transform(
    X_behavior
)


print(f"Events   : {X_behavior.shape[0]:,}")
print(f"Features : {X_behavior.shape[1]}")

print("\nBehavior features:")

for i, feature in enumerate(
    behavior_features,
    1
):
    print(f"{i:02d}. {feature}")


# ------------------------------------------------------------
# 3. Test K values
# ------------------------------------------------------------

behavior_results = []

print("\n" + "=" * 70)
print("K-MEANS MODEL COMPARISON")
print("=" * 70)

for k in range(2, 7):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = model.fit_predict(
        X_behavior_scaled
    )

    silhouette = silhouette_score(
        X_behavior_scaled,
        labels
    )

    behavior_results.append({
        "k": k,
        "inertia": model.inertia_,
        "silhouette_score": silhouette
    })

    print(
        f"K={k} | "
        f"Inertia={model.inertia_:,.2f} | "
        f"Silhouette={silhouette:.4f}"
    )


behavior_results = pd.DataFrame(
    behavior_results
)

print("\n")
display(behavior_results)


# ------------------------------------------------------------
# 4. Select best K mathematically
# ------------------------------------------------------------

best_k = int(
    behavior_results.loc[
        behavior_results["silhouette_score"].idxmax(),
        "k"
    ]
)

best_score = behavior_results[
    "silhouette_score"
].max()

print(
    f"\nBest behavior K : {best_k}"
)

print(
    f"Best silhouette : {best_score:.4f}"
)


# ------------------------------------------------------------
# 5. Train final behavior model
# ------------------------------------------------------------

behavior_kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=20
)

behavior_labels = behavior_kmeans.fit_predict(
    X_behavior_scaled
)

events_ml["behavior_cluster"] = (
    behavior_labels
)


# ------------------------------------------------------------
# 6. Cluster sizes
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BEHAVIOR CLUSTER SIZES")
print("=" * 70)

cluster_sizes = (
    events_ml["behavior_cluster"]
    .value_counts()
    .sort_index()
)

for cluster_id, count in cluster_sizes.items():

    pct = (
        count /
        len(events_ml) *
        100
    )

    print(
        f"Cluster {cluster_id}: "
        f"{count:,} events "
        f"({pct:.2f}%)"
    )


# ------------------------------------------------------------
# 7. Behavior profiles
# ------------------------------------------------------------

behavior_profile = (
    events_ml
    .groupby("behavior_cluster")
    [behavior_features]
    .mean()
    .T
)

print("\n" + "=" * 70)
print("BEHAVIOR CLUSTER PROFILES")
print("=" * 70)

display(
    behavior_profile
)


# ------------------------------------------------------------
# 8. Persistence profile
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PERSISTENCE BY BEHAVIOR CLUSTER")
print("=" * 70)

display(
    events_ml
    .groupby("behavior_cluster")
    [
        [
            "active_days",
            "duration_days",
            "activity_frequency",
            "detections_per_active_day"
        ]
    ]
    .agg(["mean", "median", "max"])
)


# ------------------------------------------------------------
# 9. Save
# ------------------------------------------------------------

behavior_results.to_csv(
    "behavior_kmeans_comparison_v10.csv",
    index=False
)

events_ml.to_csv(
    "viirs_event_behavior_clusters_v10.csv",
    index=False
)

print(
    "\nSaved:"
    "\nbehavior_kmeans_comparison_v10.csv"
    "\nviirs_event_behavior_clusters_v10.csv"
)
```
